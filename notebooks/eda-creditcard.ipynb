```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Load the credit card data
credit_df = pd.read_csv('../data/raw/creditcard.csv')

# The 'Amount' column scaling to match the PCA features (V1-V28)
scaler = StandardScaler()
credit_df['Amount'] = scaler.fit_transform(credit_df[['Amount']])

credit_df = credit_df.drop(['Time'], axis=1)

# Save
credit_df.to_csv('../data/processed/creditcard_final.csv', index=False)
print("SUCCESS: Credit card data prepared.")
```
