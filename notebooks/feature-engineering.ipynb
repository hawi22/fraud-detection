```python
import pandas as pd
import os
from sklearn.preprocessing import StandardScaler

# 1. Load the file
df = pd.read_csv('../data/processed/fraud_data_with_country.csv')

# 2. Convert time columns to datetime
df['signup_time'] = pd.to_datetime(df['signup_time'])
df['purchase_time'] = pd.to_datetime(df['purchase_time'])

# 3. FEATURE ENGINEERING
print("Engineering features...")

# Feature: Time between signup and purchase
df['time_since_signup'] = (df['purchase_time'] - df['signup_time']).dt.total_seconds()

# Feature: Transaction frequency/velocity (How many times was this device used?)
df['device_usage_count'] = df.groupby('device_id')['device_id'].transform('count')

# Feature: Time of day and Day of week
df['hour_of_day'] = df['purchase_time'].dt.hour
df['day_of_week'] = df['purchase_time'].dt.dayofweek

# 4. ONE-HOT ENCODING (Categorical to Numerical)
df = pd.get_dummies(df, columns=['source', 'browser', 'sex'], drop_first=True)

# 5. SELECTING FINAL FEATURES & SCALING
features = ['purchase_value', 'age', 'time_since_signup', 'device_usage_count', 
            'hour_of_day', 'day_of_week']

# Add the dummy columns (source, browser, sex) automatically
dummy_cols = [col for col in df.columns if 'source_' in col or 'browser_' in col or 'sex_' in col]
features.extend(dummy_cols)

# Scale numerical features so they are in the same range
scaler = StandardScaler()
df[features] = scaler.fit_transform(df[features])

# 6. SAVE FINAL VERSION
df.to_csv('../data/processed/fraud_data_final.csv', index=False)
print("SUCCESS: Task 1 Feature Engineering complete. File saved: fraud_data_final.csv")
```
